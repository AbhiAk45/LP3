```python

import threading
import time

# Take Matrix Size
n = int(input("Enter matrix size: "))

# Input Matrix A
print("Enter elements of Matrix A:")
A = []
for i in range(n):
    A.append(list(map(int, input().split())))

# Input Matrix B
print("Enter elements of Matrix B:")
B = []
for i in range(n):
    B.append(list(map(int, input().split())))

C = [[0]*n for _ in range(n)]
D = [[0]*n for _ in range(n)]

# Normal Multiplication
start = time.time()

for i in range(n):
    for j in range(n):
        for k in range(n):
            C[i][j] += A[i][k] * B[k][j]

normal_time = time.time() - start

# Multithreaded Multiplication
def multiply_row(i):
    for j in range(n):
        for k in range(n):
            D[i][j] += A[i][k] * B[k][j]

start = time.time()
threads = []

for i in range(n):
    t = threading.Thread(target=multiply_row, args=(i,))
    threads.append(t)
    t.start()

for t in threads:
    t.join()

thread_time = time.time() - start

# Display Results
print("\nNormal Matrix Multiplication:")
for row in C:
    print(row)

print("\nMultithreaded Matrix Multiplication:")
for row in D:
    print(row)

print("\nNormal Execution Time:", normal_time)
print("Multithreaded Execution Time:", thread_time)

print("\nPerformance Comparison:")
if normal_time < thread_time:
    print("Normal multiplication is faster")
else:
    print("Multithreaded multiplication is faster")
```
